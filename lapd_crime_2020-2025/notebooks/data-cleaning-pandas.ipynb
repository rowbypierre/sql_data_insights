# Import Libraries

```python
# Alter data.
import pandas as pan
# OS interactions.
import os
# Print screen alternative. 
from IPython.display import display
```

# Read CSV

```python
# Get working directory
cwd_loc = os.getcwd()
print("Working directory:\t" + cwd_loc, "\n")

data_filename = "data-raw.csv"
data_file_exist = False
cwd_filenames = list()
with os.scandir(cwd_loc + "/data") as directory:
    for element in directory:
        if element.is_file():
            cwd_filenames.append(element.name)
            print(f"Name:\t\t\t{element.name}\nPath:\t\t\t{element.path}")

            # Read file if file name found in directory.
            if element.name == data_filename:
                data_file_exist = True
                crime_data = pan.read_csv(element.path)

print(f"\nCrime data file exist:\t{data_file_exist}")
```

### 1. *Delete redundant coloumns.*

```python
# Record count
crime_data_length = len(crime_data)
print('Records #:\t' + str(crime_data_length))
```

```python
columns_to_drop = ['Vict Descent', 'Weapon Used Cd', 'Status']
crime_data_drop = crime_data.drop(columns = columns_to_drop)

print('Orginal crime data length:\t' + str(crime_data.shape[0]))
print('Orginal crime data width:\t' + str(crime_data.shape[1]))
print('Columns dropped:\t\t' + str(len(columns_to_drop)))
print('Test crime data length:\t\t' + str(crime_data_drop.shape[0]))
print('Test crime data width:\t\t' + str(crime_data_drop.shape[1]))
```

### 2. *Renaming the columns.*
- Remove spaces.
- Remove dashes.
- Add underscores.

```python
def line_break(dash_count=40):
    print('\n', '-' * dash_count, '\n')

# New capitalized column names without dashes or spaces
new_column_names = []
print("**NEW COLUMN NAMES**")
for name in crime_data_drop.columns:
    print(name.replace(' ','_').replace('-','_').upper())
    new_column_names.append(name.replace(' ','_').replace('-','_').upper())

crime_data_rename = crime_data_drop.copy() 
crime_data_rename.columns = new_column_names

line_break()
equity = False
if set(new_column_names) == set(crime_data_rename.columns):
    equity = True
print("Column names change: {}".format(equity))

line_break()
crime_data_rename.info()
```

### 3. *Drop duplicates.*

```python
# Check duplicate values
print(crime_data_rename.duplicated().value_counts())

line_break()
duplicate_values = False
if crime_data_rename.duplicated().sum() > 1:
    duplicate_values = True
print(f'Duplicate value(s) exist: {str(duplicate_values)}')
```

### 5. *Clean each column.*
- Capitalize string fields.
- Reset index.
- Convert applicable float fields to integer.

```python
crime_data_rename
```

```python
# Capitalize string fields
columns_lowercased = ["CRM_CD_DESC", "STATUS_DESC", "AREA_NAME"]
for column in columns_lowercased:
    crime_data_rename[column] = crime_data_rename[column].str.upper()
    display(crime_data_rename[column])
    line_break()
```

```python
crime_data_rename.reset_index(inplace = True)
crime_data_rename
```

```python
columns_float = list()
print("**FLOAT COLUMNS**")
for column in crime_data_rename.columns:
    if "float" in str(crime_data_rename.dtypes[column]):
        print(column)
        columns_float.append(column)

line_break()
for column in columns_float:
    display(column, crime_data_rename[column].unique())
    line_break()

columns_keep_as_float = ["LON", "LAT"]
columns_to_int = [item for item in columns_float if item not in columns_keep_as_float]
print("**FLOAT TO INT COLUMNS**")
print(columns_to_int)
    
```

```python
# Fill in nulls with '0', convert float to int
for column in columns_to_int:
    crime_data_rename[column] = crime_data_rename[column].fillna(0).astype(int)
    print("Column: {}, Data type: {} ".format(column, crime_data_rename.dtypes[column]))
```

```python
crime_data_rename.info()
```

```python
crime_data_reformat = crime_data_rename.copy()
```

```python
# Remove tab from text fields
text_field_w_tab = ["LOCATION", "CROSS_STREET"]
for field in text_field_w_tab:
    crime_data_reformat[field] = crime_data_reformat[field].str.replace("\t", " ")
    crime_data_reformat[field] = crime_data_reformat[field].str.strip()
```

### 7. *Save new data.*

```python
# Save data
file_name = "./data/data-clean-pandas.csv"
crime_data_clean = crime_data_reformat.copy()
crime_data_clean.drop(columns= ["index"], inplace= True)
crime_data_clean.to_csv(file_name, index= False)

# Read & view cleaned data
crime_data_clean = pan.read_csv(file_name)
crime_data_clean
```
